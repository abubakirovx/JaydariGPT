```python
text = "Hello world this is a simple example of how tokenization works in NLP and a sentence"
print(text)
```

```python
tokens = text.split()
print(tokens)
```

```python
vocab={}
current_id=0
for token in tokens:
  if token not in vocab:
    vocab[token] = current_id
    current_id+=1

print(vocab)
```

```python
encoded= [vocab[token] for token in tokens]
print(encoded)
```

```python
reverse_vocab={id:token for token, id in vocab.items()}
decoded=" ".join([reverse_vocab[id] for id in encoded])
print(decoded)
```

```python
import torch
import torch.nn as nn

embedding_dim = 8
vocab_size = len(vocab)

embedding_layer=nn.Embedding(vocab_size,embedding_dim)

input_ids=torch.tensor([encoded])
embeddings=embedding_layer(input_ids)

print(embeddings)
print(embeddings.shape)
```

```python
from transformers import AutoTokenizer

model_id="openai/whisper-large-v3"
tokenizer=AutoTokenizer.from_pretrained(model_id)
```

```python
print("Vocab_size:",tokenizer.vocab_size)
print("special tokens_map",tokenizer.special_tokens_map)
```

```python
text="Hello how are you doing !"
print("text=>",text)
```

```python
tokens = tokenizer.tokenize(text)
print(tokens)
```

```python
ids=tokenizer.convert_tokens_to_ids(tokens)
print(ids)
```

```python
encoded=tokenizer(text)
print(encoded)
```

```python
print(encoded["input_ids"])
```

```python
decoded_text=tokenizer.decode(encoded["input_ids"])
print(decoded_text)
```

```python
texts=[
    "Hello hi are you",
    "today we are learning about tokenizers",
    "Tiny lama is small but powerfull langauge model"
]
batch=tokenizer(texts,padding=True,return_tensors="pt")
print(batch.keys())

print("IDs shape=>",batch["input_ids"].shape)
print("attention_mask shape=>",batch["attention_mask"].shape)
```
